# ABC

```python
import lfi.inference.abc
```

```python
np.random.seed(21355)
```

```python
D = 2
budget = 1000
prior = priors.UniformPrior(low=-10, high=10, dim=D)
sim = simulators.BimodalGaussian(sigma_noise=0.1)
observation = np.zeros((1,D))
posterior_modes = np.ones((2, D))
posterior_modes[0, :] = -3
posterior_modes[1, :] = +3
```

```python
abc = lfi.inference.abc.ABCRejection(
    prior = prior,
    simulator = sim,
    observation = observation,
    eps=3)
```

```python
posterior_samples, time = abc.fit_and_sample(budget=1000, num_samples=100)
```

```python
print(f"Inference time: {time}")
```

```python
abc.plot_posterior_samples(
    samples=posterior_samples,
    budget=1000,
    posterior_modes=posterior_modes,
    subset_dims=[d for d in range(D)],
    )
```

## SBI

```python
import lfi.inference.from_sbi_abc
```

```python
import pandas as pd
import seaborn as sns
import typing
```

```python
def plot_posterior_samples(samples: np.ndarray,
                           dim: int,
                           budget: int, 
                           posterior_modes: np.ndarray = None, 
                           subset_dims: typing.Union[None, list]=None, 
                           limits = [-10, 10], 
                           name = "MCABC", 
                           savefig=None):
    samples_df = pd.DataFrame(samples, columns = [f"x_{i+1}" for i in range(samples.shape[1])])
    # Create the pairplot
    g = sns.pairplot(
    data=samples_df,
    vars = [f"x_{i+1}" for i in subset_dims],
    kind='scatter', # Base pairplot kind
    diag_kind = 'kde', # KDE for diagonal plots
    plot_kws = {'alpha':0.5}, # Scatter plot transparency)
    )

    # Create the pairplot title
    title = f"{name}, D:{dim}, budget: {budget}"

    g.fig.suptitle(title)

    # Optionally add posterior mode points
    if posterior_modes is not None:
        posterior_modes_df = pd.DataFrame(
            posterior_modes,
            columns = [f"x_{i+1}" for i in range(samples.shape[1])])
        for dim_x in subset_dims:
            for dim_y in subset_dims:
                if dim_x != dim_y: # Skip the diagonal plots
                    ax = g.axes[subset_dims.index(dim_y), subset_dims.index(dim_x)]
                    ax.scatter(
                        posterior_modes_df[f"x_{dim_x+1}"],
                        posterior_modes_df[f"x_{dim_y+1}"],
                        color='red',
                        label = 'Posterior Modes'
                    )

    # Save figure if savefig is provided
    if savefig:
        plt.savefig(savefig, bbox_inches='tight')
        
    plt.show()
    return g
    
```

```python
D = 2
prior = lfi.priors.UniformPrior(low=-10, high=10, dim=D).return_sbi_object()
sim = lfi.simulators.BimodalGaussian(sigma_noise=0.1)
observation = np.zeros((1,D))
posterior_modes = np.ones((2,D))
posterior_modes[0, :] = -3
posterior_modes[1, :] = +3
```

```python
print(type(prior))
print(type(sim))
print(type(sim.sample_pytorch))
```

```python
from sbi.inference import MCABC

inference = MCABC(prior=prior,
                  simulator=sim.sample_pytorch)
samples = inference(observation,
                    num_simulations=1_000,
                    quantile=0.1)
```

```python
g = plot_posterior_samples(
    samples=samples,
    dim=D,
    budget=1000,
    posterior_modes = posterior_modes,
    subset_dims = [d for d in range(D)],
)
```
